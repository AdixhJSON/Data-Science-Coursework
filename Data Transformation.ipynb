```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder
```

```python
# Load dataset
df = pd.read_csv("C:/Users/adish/OneDrive/Documents/adult_with_headers (1).csv")

# Display first few rows
print(df.head())
```

```python
#Basic Data Exploration
# Dataset shape
print("Shape:", df.shape)

# Data types
print(df.dtypes)

# Summary statistics
print(df.describe())

# Check missing values
print(df.isnull().sum())
```

```python
#Handle Missing Values

# Replace '?' with NaN
df.replace("?", np.nan, inplace=True)

# Check missing values again
print(df.isnull().sum())
```

```python
#Imputation
# Categorical columns
cat_cols = df.select_dtypes(include='object').columns

# Numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Impute categorical columns with mode
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)

# Impute numerical columns with median
for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
#5. Scaling Numerical Features

#5.1 Standard Scaling
standard_scaler = StandardScaler()

standard_cols = ['age', 'hours_per_week', 'education_num']
df[standard_cols] = standard_scaler.fit_transform(df[standard_cols])
```

```python
#5.2 Min-Max Scaling
minmax_scaler = MinMaxScaler()

minmax_cols = ['capital_gain', 'capital_loss']
df[minmax_cols] = minmax_scaler.fit_transform(df[minmax_cols])
```

```python
#6. Encoding Categorical Variables


#6.1 One-Hot Encoding (≤ 5 categories)

onehot_cols = ['sex', 'race']

df = pd.get_dummies(df, columns=onehot_cols, drop_first=True)
```

```python
#6.2 Label Encoding (> 5 categories)
label_cols = ['workclass', 'education', 'marital_status',
              'occupation', 'relationship', 'native_country']

le = LabelEncoder()

for col in label_cols:
    df[col] = le.fit_transform(df[col])
```

```python
#7. Feature Engineering
#7.1 Capital Net Feature
df['capital_net'] = df['capital_gain'] - df['capital_loss']
```

```python
#Work Intensity Feature
df['work_intensity'] = df['hours_per_week'] / df['age']
```

```python
#Feature Transformation (Log Transformation)
df['log_capital_gain'] = np.log1p(df['capital_gain'])
```

```python
#Encode Target Variable
df['income'] = df['income'].map({'<=50K': 0, '>50K': 1})
```
