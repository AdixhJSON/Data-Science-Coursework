```python
target=df[['Type']]
features= df.drop(columns=['Type'])
features.columns
```

```python
### feature scaling
std_sca=StandardScaler()
std_sca.fit_transform(features)
features
```

```python
plt.figure(figsize=(10,7))
features.boxplot()
plt.show()
```

```python
### outlier capping
def outlier_capping(df,column):
    Q1= df[column].quantile(0.25)
    Q3= df[column].quantile(0.75)
    IQR= Q3-Q1
    lower_extreme= Q1-1.5*IQR
    upper_extreme= Q3+1.5*IQR
    df[column]= df[column].apply(lambda x: lower_extreme if x<lower_extreme else upper_extreme if x>upper_extreme else x)
for col in features.select_dtypes(['int','float']).columns:
    outlier_capping(features,col)
```

```python
plt.figure(figsize=(10,7))
features.boxplot()
plt.show()
```

```python
x_train,x_test,y_train,y_test= train_test_split(features,target,train_size=0.75,random_state=100)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
### Random forest classifier
```

```python
rand_for= RandomForestClassifier(n_estimators=100,bootstrap=True,max_features='sqrt',
                                 max_samples=0.9,random_state=50)
```

```python
rand_for.fit(x_train,y_train)
```

```python
y_pred=rand_for.predict(x_train)
y_pred
```

```python
from sklearn.metrics import confusion_matrix,classification_report
confusion_matrix(y_train,y_pred)
```

```python
print(classification_report(y_train,y_pred))
```

```python
y_pred1=rand_for.predict(x_test)
y_pred1
```

```python
confusion_matrix(y_test,y_pred1)
```

```python
print(classification_report(y_test,y_pred1))
```
