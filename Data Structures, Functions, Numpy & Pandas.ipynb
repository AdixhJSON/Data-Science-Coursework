1. Create a list of 5 integers. Perform and print the result of the following operation: append, extend, insert, remove, pop, clear, index, count, sort, and reverse.

```python
lst = [10, 20, 30, 40, 50]

print("Original list:", lst)

lst.append(60)
print("Append:", lst)

lst.extend([70, 80])
print("Extend:", lst)

lst.insert(2, 25)
print("Insert:", lst)

lst.remove(40)
print("Remove:", lst)

lst.pop()
print("Pop:", lst)

print("Index of 30:", lst.index(30))

print("Count of 30:", lst.count(30))

lst.sort()
print("Sort:", lst)

lst.reverse()
print("Reverse:", lst)

lst.clear()
print("Clear:", lst)
```

11. Create a small DataFrame manually with 10 rows and columns: 'Name','Age','City',and 'Salary'. Then:
-Use .info(),.describe()
-Select 'Name' and 'City' columns
-Drop 'City' columns
-Fill any missing values in 'Salary' column with the mean
-Remove any duplicate rows

```python
import pandas as pd 
import numpy as np
data = {
'Name' : ['Adish', 'Rahul', 'Arjun', 'Anu' , 'Vijay', 'Neha', 'Ravi', 'Priya', 'Aman', 'Kiran'],
'Age' : [22, 25, 23, 21, 28, 24, 26, 22, 27, 25],
'City': ['Bangalore', 'Mumbai' , 'Delhi', 'Kochi', 'Chennai', 'Pune', 'Kochi', 'Delhi', 'Mumbai', 'Kochi'],
'Salary': [30000, 35000, 40000, np.nan, 45000, 32000, 38000, np.nan, 50000, 35000]
}
df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)
print("\nDataFrame Info:")
df.info()
print("\nDescription:")
print(df.describe())
print("\nName and City:")
print(df[['Name', 'City']])
df = df.drop('City', axis=1)
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df = df.drop_duplicates()
print("\nFinal DataFrame:")
print(df)
        
```

```python
# 4. Fill missing Salary values with mean
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
```
